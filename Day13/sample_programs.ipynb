```python
data={
    'pradeep':{'status':True,'python':98,'mysql':95,'flask':94},
    'siri':{'status':True,'python':78,'mysql':65,'flask':84},
    'jassu':{'status':False,'python':None,'mysql':None,'flask':None},
    'vamsi':{'status':True,'python':68,'mysql':55,'flask':64},
    'achu':{'status':True,'python':33,'mysql':25,'flask':34}
}
name=input("Enter name:")
if name in data:
  if data[name]['status']:
    total=data[name]['python']+data[name]['mysql']+data[name]['flask']
    avg=total/3
    if avg>90:
      print(f"Congratulations {name}, you got A grade")
    elif avg>70:
      print(f"Congratulations {name}, you got B grade")
    elif avg>35:
      print(f"Congratulations {name}, you got C grade")
    else:
      print(f'{name}, you have failed in the exam')
  else:
    print(f"{name} didn't write the exam")
else:
  print(f"{name} data is not found")

```
