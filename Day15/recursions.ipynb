```python
def sumofdigits(n):
  if n==0:
    return 0
  return n+sumofdigits(n-1)
print(sumofdigits(5))
```

```python
def sumofdigits(n):
  if n==0:
    return 1
  return n*sumofdigits(n-1)
print(sumofdigits(5))
```

```python
def power(bas,pow):
  if pow==0:
    return 1
  return bas*power(bas,pow-1)
print(power(2,4))
print(power(3,3))
```

```python
def reverseofstr(s,ind):
  if ind==0:
    return s[0]
  return s[ind]+reverseofstr(s,ind-1)
l="Python Programming"
print(reverseofstr(l,__builtins__.len(l)-1))
```
