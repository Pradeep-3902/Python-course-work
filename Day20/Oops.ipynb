```python
class Flipkart:
  discount=10
  products=['laptop','phone','mouse','charger']
  @classmethod
  def showproducts(cls):
    print(cls.products)

  def login(self,username,password):
    self.username=username
    self.password=password
    print(f'welcome to the flipkart {self.username}')

  @staticmethod
  def banner():
    print('10% discount is going on flipkart, shop now!')
pradeep=Flipkart()
pradeep.login('pradeep','pradeep@123')
pradeep.banner()
pradeep.showproducts()



```
