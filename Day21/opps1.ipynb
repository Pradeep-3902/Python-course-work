```python
class Instagram:
  def __init__(self):
    self._post=[]

  @property
  def accesspost(self):
    return self._post

  @accesspost.setter
  def accesspost(self,newpost):
    self._post.append(newpost)

pradeep=Instagram()

print(pradeep.accesspost)
pradeep.accesspost='class and object'
print(pradeep.accesspost)
```

```python
class whatsappv1:
  def status(self):
    print('You can upload images/videos')

class whatsappv2(whatsappv1):
  def status(self):
    super().status()
    print('You can react and reply')


class whatsappv3(whatsappv2,whatsappv1):
  def status(self):
    whatsappv1.status(self)
    whatsappv2.status(self)
    print('You can like and reshare')

kasi=whatsappv3()
kasi.status()
```
