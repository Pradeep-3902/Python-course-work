## Common String Operations in Python

Python strings are immutable sequences of characters. Here are some common operations you can perform on them:

### 1. Concatenation
You can join two or more strings using the `+` operator.

```python
str1 = "Hello"
str2 = "World"
result = str1 + " " + str2
print(result) # Output: Hello World
```

### 2. Repetition
You can repeat a string multiple times using the `*` operator.

```python
text = "Python "
repeated_text = text * 3
print(repeated_text) # Output: Python Python Python
```

### 3. Slicing
You can extract a portion of a string using slicing `[start:end:step]`.

```python
my_string = "Programming"
print(my_string[0:5])   # Output: Progr
print(my_string[5:])    # Output: amming
print(my_string[:5])    # Output: Progr
print(my_string[::2])   # Output: Prgamn
print(my_string[::-1])  # Output: gnimmargorP
```

### 4. Length
The `len()` function returns the number of characters in a string.

```python
message = "Colab"
length = len(message)
print(length) # Output: 5
```

### 5. Membership (in and not in)
You can check if a substring is present in a string.

```python
sentence = "This is a test sentence."
print("test" in sentence)     # Output: True
print("apple" not in sentence) # Output: True
```

### 6. String Methods
Strings have many built-in methods for various operations:

*   **`lower()` / `upper()`**: Converts the string to lowercase or uppercase.
    ```python
    text = "Hello World"
    print(text.lower()) # Output: hello world
    print(text.upper()) # Output: HELLO WORLD
    ```

*   **`strip()` / `lstrip()` / `rstrip()`**: Removes leading/trailing whitespace.
    ```python
    text = "  Hello  "
    print(text.strip()) # Output: Hello
    ```

*   **`replace(old, new)`**: Replaces occurrences of a substring.
    ```python
    text = "I like apples."
    new_text = text.replace("apples", "bananas")
    print(new_text) # Output: I like bananas.
    ```

*   **`split(delimiter)`**: Splits the string into a list of substrings based on a delimiter.
    ```python
    data = "apple,banana,orange"
    fruits = data.split(",")
    print(fruits) # Output: ['apple', 'banana', 'orange']
    ```

*   **`join(iterable)`**: Joins elements of an iterable (e.g., list) into a single string.
    ```python
    words = ["Hello", "Python", "!"]
    joined_string = " ".join(words)
    print(joined_string) # Output: Hello Python !
    ```

*   **`find(substring)` / `index(substring)`**: Returns the lowest index of the substring if found. `index()` raises an error if not found, `find()` returns -1.
    ```python
    text = "Python is fun"
    print(text.find("is"))   # Output: 7
    print(text.index("fun")) # Output: 10
    ```

*   **`startswith(prefix)` / `endswith(suffix)`**: Checks if the string starts or ends with a specified prefix/suffix.
    ```python
    filename = "document.pdf"
    print(filename.endswith(".pdf")) # Output: True
    ```

*   **`count(substring)`**: Returns the number of non-overlapping occurrences of substring.
    ```python
    text = "banana"
    print(text.count("a")) # Output: 3
    ```
