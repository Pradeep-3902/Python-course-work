```python
from abc import ABC, abstractmethod

class BankAccount(ABC):
    def checkbalance(self):
        print("You can checkout your balance")

    def viewhistory(self):
        print("You can your transactions")

    def userinfo(self):
        print("You can see your details")

    def transactions(self):
        print("You can transfer money through netbanking")

    @abstractmethod
    def deposit(self):
        pass

    @abstractmethod
    def withdraw(self):
        pass

class CurrentAccount(BankAccount):
    def deposit(self):
        print("You can deposit - CA")
    def withdraw(self):
        print("You can withdraw - CA")

class SavingAccount(BankAccount):
    def deposit(self):
        print("You can deposit - SA")
    def withdraw(self):
        print("You can withdraw - SA")

class FixedDepoist(BankAccount):
    def deposit(self):
        print("You can deposit - FD")
    def withdraw(self):
        print("You can withdraw - FD")

class SalaryAccount(BankAccount):
    def deposit(self):
        print("You can deposit - SAA")
    def withdraw(self):
        print("You can withdraw - SAA")

class ZeroBalanceAccount(BankAccount):
    def deposit(self):
        print("You can deposit - ZBA")
    def withdraw(self):
        print("You can withdraw - ZBA")


amathya = ZeroBalanceAccount()
amathya.deposit()
amathya.withdraw()
amathya.checkbalance()
amathya.viewhistory()
amathya.userinfo()
amathya.transactions()


ramakanth = SalaryAccount()
ramakanth.deposit()
ramakanth.withdraw()
ramakanth.checkbalance()
ramakanth.viewhistory()
ramakanth.userinfo()
ramakanth.transactions()
```

```python
from abc import ABC, abstractmethod

class TrainBooking(ABC):
    def checkavailability(self):
        print("You can check train seat availability")

    def passengerdetails(self):
        print("You can view passenger details")

    def ticketstatus(self):
        print("You can check ticket status")

    def cancelticket(self):
        print("You can cancel your ticket")

    @abstractmethod
    def bookticket(self):
        pass
    @abstractmethod
    def fare(self):
        pass


class ExpressTrain(TrainBooking):

    def bookticket(self):
        print("Ticket booked in Express Train")
    def fare(self):
        print("Fare: ₹500")


class SuperFastTrain(TrainBooking):

    def bookticket(self):
        print("Ticket booked in SuperFast Train")
    def fare(self):
        print("Fare: ₹800")


class achyuthexpress(TrainBooking):

    def bookticket(self):
        print("Ticket booked in achyuth Train")
    def fare(self):
        print("Fare: ₹1200")


class jaswantsuperfastexpress(TrainBooking):

    def bookticket(self):
        print("Ticket booked in jaswanthsuperfastexpress Train")

    def fare(self):
        print("Fare: ₹1000")


class VandeBharatTrain(TrainBooking):

    def bookticket(self):
        print("Ticket booked in Vande Bharat Train")

    def fare(self):
        print("Fare: ₹1500")


pradeep = jaswantsuperfastexpress()

pradeep.bookticket()
pradeep.fare()
pradeep.checkavailability()
pradeep.passengerdetails()
pradeep.ticketstatus()
pradeep.cancelticket()

print("---------------------------")
rahul = achyuthexpress()

rahul.bookticket()
rahul.fare()
rahul.checkavailability()
rahul.passengerdetails()
rahul.ticketstatus()
rahul.cancelticket()
```
