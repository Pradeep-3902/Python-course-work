```python
def iseven(num):
  if num%2==0:
    return f'{num}-Even number'
  else:
    return f'{num}-Odd number'
print(iseven(12))
print(iseven(13))
```

```python
def factorial(num):
  fact=1
  for i in range(1,num+1):
    fact*=i
  return fact
num=int(input('Enter a number :'))
print('Factorial:',factorial(num))
```

```python
def isprime(num):
  for i in range(2,num//2):
    if num%i==0:
      return f" {num} -Not a prime number"
  return f'{num}-Prime number'
num=int(input("Enter the number :"))
print(isprime(num))
```

```python
def display(*Names):
  print('Names: ',Names)

display('pradeep','vamsi','jaswanth','somu','prakash')
display('somu','pradeep','jaswanth')
display('vamsi','prakash')
display('pradeep')
```
